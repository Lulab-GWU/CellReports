```python
# example of how to calculate event rate (transient rate), amplitude use across the paper


import os
import sys
import numpy as np
import pandas as pd
import seaborn as sns
from pathlib import Path

REPO_ROOT = next(
    path for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (path / "pyproject.toml").is_file()
)
if str(REPO_ROOT / "src") not in sys.path:
    sys.path.insert(0, str(REPO_ROOT / "src"))
from Functions import *

import warnings
warnings.filterwarnings('ignore')
```

```python
RESULT_DIR = REPO_ROOT / "results" / "tables"
RESULT_DIR.mkdir(parents=True, exist_ok=True)

WT_list = ['NC128_WT']
```

# Call

```python
WT_df_e_list = []
WT_df_a_list = []


for i in range(len(WT_list)):
    
    mouse_name = WT_list[i]
    print("processing "+mouse_name)
    
    # set fps based on mouse_name
    if mouse_name.split('_')[0] in ['NC128','NC102', 'NC103']:
        fps = 20
    else:
        fps = 15
        
    
    # read in df
    #----------------------------------------------------------------------------------
    df_frame_label = pd.read_csv(os.path.join(REPO_ROOT, 'data', 'OFT_Data', mouse_name, 'trace_label.csv')).set_index('Frame', drop=True)
    
    df_whole = df_frame_label.copy().drop('Frame_Label', axis = 1)
    df_center = pd.read_csv(os.path.join(REPO_ROOT, 'data', 'OFT_Data', mouse_name, 'trace_center.csv')).set_index('Frame', drop=True)
    df_corner = pd.read_csv(os.path.join(REPO_ROOT, 'data', 'OFT_Data', mouse_name, 'trace_corner.csv')).set_index('Frame', drop=True)
    df_sniff = pd.read_csv(os.path.join(REPO_ROOT, 'data', 'OFT_Data', mouse_name, 'trace_sniff.csv')).set_index('Frame', drop=True)
    df_groom = pd.read_csv(os.path.join(REPO_ROOT, 'data', 'OFT_Data', mouse_name, 'trace_groom.csv')).set_index('Frame', drop=True)

    

    # calculate
    #----------------------------------------------------------------------------------
    whole_eventrate, whole_amplitude, _  = event_rate(df_whole, fps)
    whole_corrs = corr_list(df_whole)
    
    center_eventrate, center_amplitude, _  = event_rate(df_center, fps)
    center_corrs = corr_list(df_center)
    
    corner_eventrate, corner_amplitude, _  = event_rate(df_corner, fps)
    corner_corrs = corr_list(df_corner)
    
    snif_eventrate, snif_amplitude, _  = event_rate(df_sniff, fps)
    snif_corrs = corr_list(df_sniff)
    
    groom_eventrate, groom_amplitude, _  = event_rate(df_groom, fps)
    groom_corrs = corr_list(df_groom)

    WT_dic_e = {'mouse_name':mouse_name,
                'whole_seesion': whole_eventrate,
                'center_seesion': center_eventrate,
                'corner_seesion': corner_eventrate,
                'snif_seesion': snif_eventrate,
                'groom_seesion': groom_eventrate}
    WT_df_e = pd.DataFrame.from_dict(WT_dic_e, orient='index').T
    WT_df_e_list.append(WT_df_e)

    WT_dic_a = {'mouse_name':mouse_name,
                'whole_seesion': whole_amplitude,
                'center_seesion': center_amplitude,
                'corner_seesion': corner_amplitude,
                'snif_seesion': snif_amplitude,
                'groom_seesion': groom_amplitude}
    WT_df_a = pd.DataFrame.from_dict(WT_dic_a, orient='index').T
    WT_df_a_list.append(WT_df_a)


WT_df_ee = pd.concat(WT_df_e_list)
WT_df_ee.to_excel(RESULT_DIR / 'OFT_WT_All_Eventrate.xlsx')

WT_df_aa = pd.concat(WT_df_a_list)
WT_df_aa.to_excel(RESULT_DIR / 'OFT_WT_All_Amplitude.xlsx')
```
